```python
import tensorflow as tf

MODEL_PATH = 'Saved_Model/final_female_model.keras'
OUTPUT_PATH = 'Saved_Model/final_female_model.tflite'

# 1. Load model
model = tf.keras.models.load_model(MODEL_PATH)

# 2. Setup Converter
converter = tf.lite.TFLiteConverter.from_keras_model(model)

# --- BAGIAN KRUSIAL UNTUK LSTM/BiLSTM ---
# Mengaktifkan Select TF Ops
converter.target_spec.supported_ops = [
    tf.lite.OpsSet.TFLITE_BUILTINS, # Operasi standar TFLite
    tf.lite.OpsSet.SELECT_TF_OPS    # Operasi tambahan dari TensorFlow
]

# Mematikan penurunan otomatis tensor list (sesuai saran error)
converter._experimental_lower_tensor_list_ops = False
# ----------------------------------------

converter.optimizations = [tf.lite.Optimize.DEFAULT]

try:
    print("Sedang mengonversi model BiLSTM... (ini mungkin butuh waktu lebih lama)")
    tflite_model = converter.convert()

    # 3. Simpan hasil konversi
    with open(OUTPUT_PATH, 'wb') as f:
        f.write(tflite_model)

    print(f"✅ Berhasil! Model disimpan ke: {OUTPUT_PATH}")
    print(f"Ukuran file: {len(tflite_model) / 1024:.2f} KB")

except Exception as e:
    print(f"❌ Masih gagal: {e}")
```
